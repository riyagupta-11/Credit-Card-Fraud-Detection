```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("creditcard.csv")
df.head()
```

```python
df.drop("Time",  axis = 1)
```

```python
df.isnull().sum()
```

```python
x = df.drop("Class", axis = 1)
y = df["Class"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42, stratify = y) 
```

```python
x_train.shape , y_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn. ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(n_estimators = 50, random_state = 42, max_features = "sqrt", oob_score = True, max_depth = 3)
rf.fit(x_train_scaled, y_train)
```

```python
y_test_pred = rf.predict(x_test) 
```

```python
from sklearn. metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve
```

```python
print(accuracy_score(y_test, y_test_pred))
```

```python
cm = confusion_matrix(y_test, y_test_pred)
print(cm)
```

```python
plt.figure(figsize = (5,5))
sns.heatmap(cm, annot = True, fmt = "d", cmap = "Blues")
plt.xlabel("Features")
plt.ylabel("prediction")
plt.title("Random Forest Classifier Confusion Matrix")
plt.show()
```

```python
print(classification_report(y_test, y_test_pred))
```

```python
y_test_pred_proba = rf.predict_proba(x_test)[:, 1]
```

```python
roc_auc = roc_auc_score(y_test, y_test_pred_proba)
roc_auc
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_test_pred_proba)
```

```python
plt.figure(figsize = (8,6))
plt.plot(fpr, tpr, color = "blue", lw= 2, label = (f"roc_curve:{roc_auc}"))
plt.plot([0,1], [0,1], lw = 2, linestyle = "--", color = "red", label = "Random Guessing")
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("Features")
plt.ylabel("Predictions")
plt.title("Credit Card Amount Detection")
plt.legend(loc= "lower right")
plt.show()
```

```python
importance = rf.feature_importances_
features = x.columns 
```

```python
plt.figure(figsize= (8,6))
sns.barplot(x= importance, y= features)
plt.ylabel("Transactions")
plt.title("Random Forest Feature Classification")
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
```

```python
base_learners = [
    ("DT", DecisionTreeClassifier(random_state= 42)),
    ("knn", KNeighborsClassifier()), 
    ("svm", SVC(probability = True, random_state = 42))
]
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
meta_model = LogisticRegression()
```

```python
from sklearn.ensemble import StackingClassifier
```

```python
stack_model = StackingClassifier(estimators = base_learners, final_estimator = meta_model, cv =5)
```

```python
stack_model.fit(x_train_scaled, y_train)
```

```python
y_pred_stack = stack_model.predict(x_test_scaled)
```

```python
print(accuracy_score(y_test, y_pred_stack))
```

```python
print(classification_report(y_test, y_pred_stack))
```

```python
y_pred_proba_stack = stack_model.predict_proba(x_test_scaled)[:,1]
```

```python
roc_auc_stack = roc_auc_score(y_test, y_pred_proba_stack)
print(roc_auc_stack)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba_stack)
```
